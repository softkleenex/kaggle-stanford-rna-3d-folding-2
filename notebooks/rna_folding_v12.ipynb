```python
import os, glob, subprocess, sys
print("Installing wheels...")
wheels = glob.glob("/kaggle/input/**/*.whl", recursive=True)
for wheel in wheels:
    if "biopython" in wheel or "biotite" in wheel or "rdkit" in wheel or "biotraj" in wheel:
        print(f"Installing {os.path.basename(wheel)}...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", wheel, "--no-index"])

import Bio, biotite, rdkit
print("Verification: Biopython, Biotite, RDKit imported!")
```
